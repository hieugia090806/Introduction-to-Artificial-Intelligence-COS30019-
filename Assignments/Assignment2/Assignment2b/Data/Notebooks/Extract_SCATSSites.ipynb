```python
#--Import curcial libraries. --#
import os
import re
import pandas as pd
```

```python
input_file = '../Raw/SCATSiteNumbers.csv'
output_dir = '../Processed'
```

```python
df = pd.read_csv(input_file, sep='\t', skiprows=9)
df.columns = df.columns.str.strip()
```

```python
df['Directory'] = df['Directory'].fillna('Unknown')
df['Map reference'] = df['Map reference'].fillna('Unknown')
```

```python
if 'Site Type' in df.columns:
    # Lấy danh sách các loại Site Type duy nhất
    unique_types = df['Site Type'].dropna().unique()
    
    print(f"Found {len(unique_types)} types. Splitting files into '{output_dir}'...")

    for s_type in unique_types:
        # Lọc dữ liệu theo từng loại
        filtered_df = df[df['Site Type'] == s_type]
        
        # Tạo tên file an toàn (Ví dụ: 'FIRE W/W' -> 'FireWw')
        # Dùng regex để loại bỏ các ký tự đặc biệt như dấu /
        clean_label = "".join(word.capitalize() for word in re.sub(r'[^a-zA-Z0-9]', ' ', str(s_type)).split())
        file_name = f"SCATSType{clean_label}.csv"
        
        # Nối đường dẫn vào folder Processed
        save_path = os.path.join(output_dir, file_name)
        
        # Xuất file CSV
        filtered_df.to_csv(save_path, index=False)
        print(f"Successfully created: {file_name} with {len(filtered_df)} rows.")
else:
    print("Error: Could not find 'Site Type' column. Please check your CSV header.")

print("\n--- ALL FILES HAVE BEEN SPLIT AND SAVED ---")
```
