```python
#--Import libraries. --#
import os
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
processed_dir = '../Processed/'
figures_dir = '../Figures/'
```

```python
df = pd.read_csv('../Processed/roads_final_cleaned.csv')
df['aadt_allve'] = pd.to_numeric(df['aadt_allve'], errors='coerce')
df['last_year'] = pd.to_numeric(df['last_year'], errors='coerce')
df['road_name'] = df['declared_r'].fillna('Unknown Road')
```

```python
df_clean = df.dropna(subset=['aadt_allve', 'last_year', 'x', 'y'])
df_clean = df_clean[df_clean['aadt_allve'] > 0]
```

```python
road_avgs = df_clean.groupby('road_name')['aadt_allve'].mean().reset_index()
road_avgs.columns = ['road_name', 'road_avg_aadt']
```

```python
df_efficiency = pd.merge(df_clean, road_avgs, on='road_name')
df_efficiency['efficiency_index'] = (df_efficiency['aadt_allve'] / df_efficiency['road_avg_aadt']).round(2)
```

```python
efficiency_stations = df_efficiency.sort_values(
    ['road_name', 'efficiency_index', 'last_year'], 
    ascending=[True, False, False]
).groupby('road_name').head(3)
```

```python
csv_output_path = os.path.join(processed_dir, 'efficiency_stations.csv')
efficiency_stations.to_csv(csv_output_path, index=False)
print(f"✅ Saved file CSV at: {csv_output_path}")
```

```python
sns.set_theme(style="whitegrid")
top_showcase = efficiency_stations.sort_values('efficiency_index', ascending=False).head(10)
```

```python
for idx, row in top_showcase.iterrows():
    plt.figure(figsize=(10, 6))
    
    #-- Compare data. --#
    compare_data = pd.DataFrame({
        'Category': ['This Station', 'Road Average'],
        'AADT': [row['aadt_allve'], row['road_avg_aadt']]
    })
    
    ax = sns.barplot(x='Category', y='AADT', data=compare_data, palette=['#4169E1', '#A9A9A9'])

    for p in ax.patches:
        ax.annotate(f'{int(p.get_height())}', (p.get_x() + p.get_width() / 2., p.get_height()), 
                    ha='center', va='center', fontsize=12, color='black', xytext=(0, 5),
                    textcoords='offset points')

    plt.title(f"Efficiency Analysis: Station {row['tfm_id']} on {row['road_name']}\n"
              f"Performance: {row['efficiency_index']}x higher than road average", fontsize=14)
    plt.tight_layout()
    
    #-- Save file. --#
    img_name = f"efficiency_analysis_sid_{row['tfm_id']}.png"
    plt.savefig(os.path.join(figures_dir, img_name))
    plt.close()

print(f"✅ Save analysis figures in folder: {figures_dir}")
```
