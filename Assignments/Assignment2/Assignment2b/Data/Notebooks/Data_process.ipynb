```python
#-- Import crucial libraries. --#
import os
import glob
import numpy as np
import pandas as pd
```

```python
input_dir = '../Processed/'
output_folder = '../Processed'

file_pattern = os.path.join(input_dir, 'roads_*.csv')
all_files = glob.glob(file_pattern)

if not all_files:
    print(f"Error: No files found in {input_dir}")
else:
    df_list = []
    for f in all_files:
        temp_df = pd.read_csv(f, low_memory=False)
        df_list.append(temp_df)
    
    df = pd.concat(df_list, ignore_index=True)

    #-- DATA CLEANING STRATEGY. --#
    garbage_labels = ['Missing Data', 'unknown', 'Unknown', 'nan', ' ', '0']
    df = df.replace(garbage_labels, np.nan)

    df['aadt_allve'] = pd.to_numeric(df['aadt_allve'], errors='coerce')

    df_cleaned = df.dropna(subset=['x', 'y', 'aadt_allve'])

    if 'declared_r' in df_cleaned.columns:
        df_cleaned['declared_r'] = df_cleaned['declared_r'].fillna('Not Available')

    output_path = os.path.join(output_folder, 'roads_final_cleaned.csv')
    df_cleaned.to_csv(output_path, index=False)
```
