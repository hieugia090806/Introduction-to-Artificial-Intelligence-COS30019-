After identifying missing data, our team decide to utilize different techniques for dealing with. Secondly, this file use mean method.

```python
#-- Import crucila library. --#
import os
import glob
import numpy as np
import pandas as pd
```

```python
input_folder = '../Processed'
all_files = glob.glob(os.path.join(input_folder, 'roads_*.csv'))
df = pd.concat([pd.read_csv(f) for f in all_files], ignore_index=True)
```

```python
#-- Strategy. --#
df_mean = df.copy()
df_mean['aadt_allve'] = pd.to_numeric(df_mean['aadt_allve'], errors='coerce')

# Calculate mean and fill
avg_traffic = df_mean['aadt_allve'].mean()
df_mean['aadt_allve'] = df_mean['aadt_allve'].fillna(avg_traffic)
```

```python
#-- Saved the output. --#
output_dir = '../Processed/'
if not os.path.exists(output_dir): os.makedirs(output_dir)
df_mean.to_csv(os.path.join(output_dir, 'roads_mean_imputed.csv'), index=False)
```

```python
print(f"--- MEAN METHOD COMPLETE ---")
print(f"Filled missing values with global average: {avg_traffic:.2f}")
```
