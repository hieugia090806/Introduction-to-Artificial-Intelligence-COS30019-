```python
#---Import crucial libraries. --#
import pandas as pd
```

```python
input_path = '../Raw/DataInformation.csv'
output_path = '../Processed/Symbols.csv'
```

```python
with open(input_path, 'r', encoding='utf-8') as f:
    lines = f.readlines()
```

```python
start_idx = 0
for i, line in enumerate(lines):
    if "Site Type" in line:
        start_idx = i + 1
        break

#-- Read data with Pandas library by using Tab \t
df_symbols = pd.read_csv(input_path, sep='\t', skiprows=start_idx, header=None)
df_symbols = df_symbols.iloc[:, :2] 
df_symbols.columns = ['Symbols', 'Meaning']
df_symbols = df_symbols.dropna()

df_symbols.loc[df_symbols['Symbols'] == 'FIRE W/W', 'Meaning'] = 'Fire Station Exit (Warning Way)'
df_symbols.loc[df_symbols['Symbols'] == 'FIRE SIG', 'Meaning'] = 'Fire Station Exit (Signal Controlled)'
df_symbols.loc[df_symbols['Symbols'] == 'AMBU SIG', 'Meaning'] = 'Ambulance Exit (Signal Controlled)'

#-- Export file Symbols.csv. --#
df_symbols.to_csv(output_path, index=False)

#-- Check in terminal. --#
print("--- SYMBOLS TABLE (UPDATED FOR AI) ---")
print(df_symbols[df_symbols['Symbols'].str.contains('FIRE|AMBU')]) #-- Print special rows. --#
print("\nFull table saved to Processed/Symbols.csv")
```
