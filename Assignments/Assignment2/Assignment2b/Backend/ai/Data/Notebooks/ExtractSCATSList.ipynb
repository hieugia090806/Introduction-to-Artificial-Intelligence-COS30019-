```python
import os
import pandas as pd
```

```python
current_path = os.getcwd()
parent_dir = os.path.dirname(current_path)

# Đường dẫn chính xác nè Hieu
input_path = os.path.join(parent_dir, 'Raw', 'SCATSiteNumbers.csv')
output_dir = os.path.join(parent_dir, 'Processed')

if os.path.exists(input_path):
    print(f"✅ Have seen file at: {input_path}")
else:
    print(f"❌ Not seen! Check again)")
    # In ra để tự kiểm tra
    print(f"Parents folder has: {os.listdir(parent_dir)}")
```

```python
try:
    df_raw = pd.read_csv(input_path, sep='\t', skiprows=8)
    
    df_raw.columns = df_raw.columns.str.strip()
    
    print("📊 Columns found:", df_raw.columns.tolist())
    display(df_raw.head())
except Exception as e:
    print(f"❌ Error file: {e}")
```

```python
import os

if not os.path.exists(output_dir):
    os.makedirs(output_dir)

try:
    df_raw.columns = df_raw.columns.str.strip()
    
    print("Columns:", df_raw.columns.tolist())
    
    # 3. Trích xuất dựa trên vị trí cột (Cột 0 là ID, Cột 1 là Name) 
    # Cách này an toàn hơn là dùng tên trực tiếp
    df_clean = df_raw.iloc[:, [0, 1]].copy()
    
    # 4. Đổi tên cho chuẩn
    df_clean.columns = ['SCATS_ID', 'Station_Name']
    
    # 5. Xử lý dữ liệu
    df_clean = df_clean.dropna(subset=['SCATS_ID'])
    # Chuyển ID về dạng chuỗi sạch (không có .0)
    df_clean['SCATS_ID'] = pd.to_numeric(df_clean['SCATS_ID'], errors='coerce').fillna(0).astype(int).astype(str)
    
    # Loại bỏ các dòng tiêu đề trùng lặp nếu có
    df_clean = df_clean[df_clean['SCATS_ID'] != '0']

    # 6. Lưu file
    output_file = os.path.join(output_dir, 'stations_cleaned.csv')
    df_clean.to_csv(output_file, index=False)
    
    print(f"🎉 Success! Saved {len(df_clean)} stations.")
    print(f"💾 File: {output_file}")
    display(df_clean.head())

except Exception as e:
    print(f"❌ Error: {e}")
```
